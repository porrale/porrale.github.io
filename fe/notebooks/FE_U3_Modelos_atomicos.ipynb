```python
#@title Modelo Rutherford
import numpy as np
import matplotlib.pyplot as plt

# Configuración de la figura y fondo limpio
fig, ax = plt.subplots(figsize=(7, 7), dpi=150)
ax.set_aspect('equal')
ax.axis('off')  # Ocultamos los ejes numéricos tradicionales

# 1. Ejes de referencia punteados (Líneas azul claro)
ax.axhline(0, color='#a0c4de', linestyle='--', linewidth=1, zorder=1)
ax.axvline(0, color='#a0c4de', linestyle='--', linewidth=1, zorder=1)

# 2. Órbita circular punteada (Marrón/Dorado)
r_val = 3.0  # Radio de la órbita
theta_orbit = np.linspace(0, 2*np.pi, 200)
x_orbit = r_val * np.cos(theta_orbit)
y_orbit = r_val * np.sin(theta_orbit)
ax.plot(x_orbit, y_orbit, color='#a68953', linestyle='--', linewidth=1.5, zorder=2)

# 3. Línea de referencia del radio (Línea fina auxiliar al electrón)
angle = np.deg2rad(20)  # Ángulo de posición del electrón (~20 grados)
x_e = r_val * np.cos(angle)
y_e = r_val * np.sin(angle)
ax.plot([0, x_e], [0, y_e], color='#a0c4de', linestyle=':', linewidth=1, zorder=1)

# 4. Núcleo central (Círculo grande verde oscuro)
ax.plot(0, 0, marker='o', color='#0c4015', markersize=14, zorder=5)
ax.text(0, -0.4, r'$Z_e$', fontsize=16, ha='center', va='top', fontname='serif')

# 5. Electrón (Punto rojo en la órbita)
ax.plot(x_e, y_e, marker='o', color='#cc0000', markersize=6, zorder=5)
ax.text(x_e + 0.15, y_e, r'$e^-$', fontsize=16, ha='left', va='center')

# 6. Vector Fuerza Centrípeta (Flecha roja hacia el centro)
# El origen es el electrón (x_e, y_e) y apunta hacia el centro (dx, dy)
dx_f = -1.2 * np.cos(angle)
dy_f = -1.2 * np.sin(angle)
ax.annotate('', xy=(x_e + dx_f, y_e + dy_f), xytext=(x_e, y_e),
            arrowprops=dict(arrowstyle="->", color='#cc0000', lw=2.5, mutation_scale=15), zorder=4)
ax.text(x_e + dx_f*0.5, y_e + dy_f*0.5 + 0.25, r'$F_c$', color='black', fontsize=15, ha='center')

# 7. Vector Velocidad Tangencial (Flecha negra perpendicular al radio)
# La dirección tangencial es (-sin(angle), cos(angle))
dx_v = -1.5 * np.sin(angle)
dy_v = 1.5 * np.cos(angle)
ax.annotate('', xy=(x_e + dx_v, y_e + dy_v), xytext=(x_e, y_e),
            arrowprops=dict(arrowstyle="->", color='black', lw=1.8, mutation_scale=15), zorder=4)
ax.text(x_e + dx_v + 0.1, y_e + dy_v + 0.1, r'$v$', color='black', fontsize=15, ha='left', va='bottom')

# 8. Etiqueta del radio 'r'
ax.text(x_e * 0.4, y_e * 0.4 - 0.2, r'$r$', fontsize=14, ha='center', va='center')

# Ajustes de límites para encuadrar la órbita
limit = r_val + 1.5
ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)

# Mostrar el gráfico en Colab
plt.show()
```

```python
#@title Simulador de Onda Senoidal { run: "auto" }
#@markdown Modificá los parámetros para ver cómo cambia la señal en el tiempo.

frecuencia = 5 #@param {type:"slider", min:1, max:20, step:1}
amplitud = 2 #@param {type:"slider", min:0.5, max:5, step:0.5}

import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)
y = amplitud * np.sin(2 * np.pi * frecuencia * t)

plt.figure(figsize=(8, 4))
plt.plot(t, y, color="red", linewidth=2)
plt.title(f"Señal de {frecuencia} Hz")
plt.grid(True)
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.show()
```
